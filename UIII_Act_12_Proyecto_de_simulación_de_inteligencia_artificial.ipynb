```python
# =========================================================
# IA DE YOUTUBERS - PROGRAMACION ORIENTADA A OBJETOS
# CBTIS 128 - Programacion e Inteligencia Artificial
# =========================================================

print("===================================================")
print("        IA DE YOUTUBERS VIRTUALES")
print("      Programacion Orientada a Objetos")
print("===================================================")


# =========================================================
# CLASE YOUTUBER
# =========================================================
class Youtuber:

    # Constructor
    def __init__(self, nombre, categoria, suscriptores,
                 placas, software, camara, patrocinador):

        self.nombre = nombre
        self.categoria = categoria
        self.suscriptores = suscriptores
        self.placas = placas
        self.software = software
        self.camara = camara
        self.patrocinador = patrocinador

    # =====================================================
    # FUNCION AGREGAR YOUTUBER
    # =====================================================
    @staticmethod
    def agregar_youtuber(lista):

        print("\n========== AGREGAR YOUTUBER ==========")

        nombre = input("Nombre: ")
        categoria = input("Categoria de videos: ")
        suscriptores = input("Cantidad de suscriptores: ")
        placas = input("Placas obtenidas: ")
        software = input("Software de edicion: ")
        camara = input("Camara utilizada: ")
        patrocinador = input("Patrocinadores: ")

        nuevo = Youtuber(
            nombre,
            categoria,
            suscriptores,
            placas,
            software,
            camara,
            patrocinador
        )

        lista.append(nuevo)

        print("\n✅ Youtuber agregado correctamente.")

    # =====================================================
    # FUNCION LISTAR YOUTUBERS
    # =====================================================
    @staticmethod
    def listar_youtuber(lista):

        print("\n================ LISTA DE YOUTUBERS ================")

        if len(lista) == 0:
            print("No hay youtubers registrados.")

        else:

            print("-" * 135)

            print(
                f"{'NOMBRE':15}"
                f"{'CATEGORIA':18}"
                f"{'SUSCRIPTORES':15}"
                f"{'PLACAS':12}"
                f"{'SOFTWARE':18}"
                f"{'CAMARA':20}"
                f"{'PATROCINADOR':20}"
            )

            print("-" * 135)

            for y in lista:

                print(
                    f"{y.nombre:15}"
                    f"{y.categoria:18}"
                    f"{y.suscriptores:15}"
                    f"{y.placas:12}"
                    f"{y.software:18}"
                    f"{y.camara:20}"
                    f"{y.patrocinador:20}"
                )

            print("-" * 135)

    # =====================================================
    # FUNCION ACTUALIZAR YOUTUBER
    # =====================================================
    @staticmethod
    def actualizar_youtuber(lista):

        nombre_buscar = input("\nNombre del youtuber a actualizar: ")

        encontrado = False

        for y in lista:

            if y.nombre.lower() == nombre_buscar.lower():

                print("\n======= NUEVOS DATOS =======")

                y.categoria = input("Nueva categoria: ")
                y.suscriptores = input("Nuevos suscriptores: ")
                y.placas = input("Nuevas placas: ")
                y.software = input("Nuevo software: ")
                y.camara = input("Nueva camara: ")
                y.patrocinador = input("Nuevo patrocinador: ")

                print("\n✅ Datos actualizados correctamente.")

                encontrado = True

        if encontrado == False:
            print("❌ Youtuber no encontrado.")

    # =====================================================
    # FUNCION BORRAR YOUTUBER
    # =====================================================
    @staticmethod
    def borrar_youtuber(lista):

        nombre_buscar = input("\nNombre del youtuber a borrar: ")

        encontrado = False

        for y in lista:

            if y.nombre.lower() == nombre_buscar.lower():

                lista.remove(y)

                print("\n🗑️ Youtuber eliminado correctamente.")

                encontrado = True

        if encontrado == False:
            print("❌ Youtuber no encontrado.")


# =========================================================
# LISTA PRINCIPAL
# =========================================================
lista_youtubers = []


# =========================================================
# MENU PRINCIPAL
# =========================================================
while True:

    print("\n================ MENU PRINCIPAL ================")
    print("1. Agregar youtuber")
    print("2. Listar youtubers")
    print("3. Actualizar youtuber")
    print("4. Borrar youtuber")
    print("5. Salir")

    opcion = input("Selecciona una opcion: ")

    # =====================================================
    # OPCIONES DEL MENU
    # =====================================================

    if opcion == "1":
        Youtuber.agregar_youtuber(lista_youtubers)

    elif opcion == "2":
        Youtuber.listar_youtuber(lista_youtubers)

    elif opcion == "3":
        Youtuber.actualizar_youtuber(lista_youtubers)

    elif opcion == "4":
        Youtuber.borrar_youtuber(lista_youtubers)

    elif opcion == "5":
        print("\n👋 Cerrando sistema IA de Youtubers...")
        break

    else:
        print("❌ Opcion invalida.")
```
